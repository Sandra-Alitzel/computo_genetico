```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

# Cargar datos
df = pd.read_csv('symbolic_regression_2.csv')

# Aquí pondrías la expresión de tu último registro en gp_history.json
# Nota: Hay que traducir 'div(a, b)' a '(a/b)' para que Python lo entienda
def modelo_gp(x, y):
    # Ejemplo simplificado de lo que hay en tu JSON:
    return np.sin(x) * (y / 4.4) 

df['Z_pred'] = modelo_gp(df['X'], df['Y'])
df['Error'] = np.abs(df['Z'] - df['Z_pred'])

# Visualización
fig, ax = plt.subplots(1, 2, figsize=(15, 6))

# Mapa de Calor del Error
sc = ax[0].scatter(df['X'], df['Y'], c=df['Error'], cmap='Reds')
plt.colorbar(sc, ax=ax[0], label='Error Absoluto')
ax[0].set_title('¿Dónde falla el modelo? (Distribución del Error)')

# Actual vs Predicted
ax[1].scatter(df['Z'], df['Z_pred'], alpha=0.5)
ax[1].plot([df['Z'].min(), df['Z'].max()], [df['Z'].min(), df['Z'].max()], 'r--')
ax[1].set_title('Ajuste: Real vs Predicho')
ax[1].set_xlabel('Real')
ax[1].set_ylabel('Predicho')

plt.show()
```
